```python
from datetime import datetime


def _dt_to_ts(dt: datetime | None) -> Timestamp | None:
    print(dt)
    if dt is None:
        print("dt is none")
        return None
    ts = Timestamp().FromDatetime(dt)
    print(f"ts: {ts}")
    return ts
```
